1. Дослідимо, який філіал супермаркету ('Branch') є найприбутковішим. Для цього знайдіть сумарний прибуток за кожним філіалом і виявіть, який філіал має найвищий.

```python
profit_by_filial = df.groupby('Branch')['gross income'].sum()
print(profit_by_filial)


top_profitable = profit_by_filial.idxmax()
print("Найприбутковіший філіал:", top_profitable)
```

2. В якому місті знайходиться філіал з найвищим прибутком? Може в тому місці нам розмітисти ще один магазин.  
Знайдіть відповідь за допомогою функціоналу Pandas.

```python
pivot_city = df.pivot_table(
    index='Branch',
    values=['gross income', 'City'],
    aggfunc={'gross income': 'sum', 'City': 'first'}
)

best_filial = pivot_city['gross income'].idxmax()
city_top_profitable = pivot_city.loc[best_filial, 'City']

city_top_profitable
```

3.1. Створіть зводну таблицю, яка покаже, скільки покупок (інвойсів) було зроблено в кожній з філій (`Branch`) за різними категоріями товарів. Запишіть таблицю в змінну `invoices_by_category` і виведіть змінну на екран.
Ця таблиця допоможе проаналізувати, в якій філії купують найбільше товарів кожної з категорій.

```python
invoices_by_category = df.pivot_table(
    index='Branch',
    columns='Product line',
    values='Invoice ID',
    aggfunc='count'
)
print(invoices_by_category)
```

```python
#Відповідь на питання №1

avg_total_gender = df.groupby('Gender')['Total'].mean()
print(avg_total_gender)

top_spenders = avg_total_gender.idxmax()
print("Хто витрачає більше за один раз:", top_spenders)
```

```python
#Відповідь на питання №2
popular_product_by_city = df.groupby(['City', 'Product line'])['Invoice ID'].count()
popular_product_by_city = popular_product_by_city.groupby('City').idxmax()


print(popular_product_by_city)
```

```python
#Відповідь на питання №3
avg_price_by_city = df.groupby('City')['Unit price'].mean()
print(avg_price_by_city)

city_with_expensive = avg_price_by_city.idxmax()
print("Місто з найдорожчими товарами:", city_with_expensive)
```
